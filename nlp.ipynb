```python
import pandas as pd
```

```python
df = pd.read_csv("all_data.csv", encoding="ISO-8859-1")
```

```python
import re 
```

```python
cleaned = [] 
```

```python
for text in df['Text']:  
    cleaned_text = re.sub(r'[^\w\s]', '', text) 
    cleaned_text = re.sub(r'\s+', ' ', cleaned_text)  
    cleaned_data = cleaned_text.strip() 
    cleaned.append(cleaned_data) 
    
```

```python
tokens = [word_tokenize(x) for x in cleaned]
```

```python
stop = set(stopwords.words('english')) 
```

```python
stpktn = []
```

```python
for k in range(len(df['Text'])):  
    p = [i for i in tokens[k] if i not in stop]  
    stpktn.append(p)
```

```python
from nltk.stem import PorterStemmer
ps = PorterStemmer()
```

```python
stemed_data = []
for Text in stpktn:
  stm = [ps.stem(word) for word in Text]
  stemed_data.append(stm)
```

```python
stemed_data
```

```python
stm=[" ".join(i) for i in stemed_data]
```

```python
from sklearn.feature_extraction.text import CountVectorizer
```

```python
cv=CountVectorizer(max_features=2000)
```

```python
vc=cv.fit(stm)
```

```python
import pickle
with open("vectorizer.pickle","wb") as vcfile:
    pickle.dump(vc,vcfile)
```

```python
x=vc.transform(stm).toarray()
```

```python
x
```

```python
y=df["Sentiment"]
```

```python
from sklearn.naive_bayes import MultinomialNB
```

```python
mb=MultinomialNB()
```

```python
mb.fit(x,y)
```
